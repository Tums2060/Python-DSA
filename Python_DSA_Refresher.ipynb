1. Write a program that asks the user for a number of days. The program then
prints out the number of seconds in the amount of days given

```python
days = int(input("What are the number of days?"))

# To get total seconds I multiply the number of days to 24 which will give me hours then to 60 to give me minutes then 60 again to give me seconds
total_seconds = days*24*60*60
print(total_seconds)
```

3. Using functions, write a program to compute the area and perimeter of a
square. The program should ask the user to enter a number corresponding to
the side length of the square and display the area and perimeter of the
square.

```python
def area_perimeter():
    side_length = int(input("What is the length of your square in cm? "))

    # area is length squared
    area = side_length * side_length
    
    # perimeter is length x 4
    perimeter = side_length * 4
    return print(f"Your square of length {side_length}, has an area of {area} cm sqrd and a perimeter of {perimeter} cm")

def main():
    output = area_perimeter()

if __name__ == "__main__":
    main()
```

4. Write a program using functions that determines whether a character input
by a user is uppercase or lower case.

```python
def upper_lowecase():
    user_character = input("What is your character? ")
    #I used the inbuilt python function "isupper"
    output = user_character.isupper()

    # The function outputs True if the character is actually uppercase, so here, if output also means if true
    if output:
        print("Your character is Upper case")
    else:
        print("Your character is Lower case")

def main():
    result = upper_lowecase()

if __name__ == "__main__":
    main()
```

5. The following is pseudocode for a program being designed. Write the
Python program equivalent of the same.

```python
def loop_function():
    x = 0
    y = 20

    # I used a while statement to loop through the addition and subtraction
    while True:
        y = y - 4
        x = x + (2/y)

        # This is the condition to stop the while loop
        if y < 6:
            break

    print(x)

def main():
    output = loop_function()

if __name__ == "__main__":
    main()
```
