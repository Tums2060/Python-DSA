1. Write a program that asks the user for a number of days. The program then
prints out the number of seconds in the amount of days given

```python
days = int(input("What are the number of days?"))

# To get total seconds I multiply the number of days to 24 which will give me hours then to 60 to give me minutes then 60 again to give me seconds
total_seconds = days*24*60*60
print(total_seconds)
```
